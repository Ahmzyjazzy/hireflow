## 2. Configure Gemini API Key

HireFlow uses Google's Gemini models. You need an API key to proceed.

1.  Get your key from [Google AI Studio](https://aistudio.google.com/app/api-keys).
2.  Run the cell below and paste your key when prompted.

```python
import os
import getpass

# Prompt for API Key
api_key = getpass.getpass("Enter your Google Gemini API Key: ")

# Set it as an environment variable
os.environ["GOOGLE_API_KEY"] = api_key

# Also write it to .env for the subprocesses to pick up
with open(".env", "w") as f:
    f.write(f"GOOGLE_API_KEY={api_key}\n")

print("API Key configured successfully!")
```

## 3. Start Services

HireFlow consists of two main parts that need to run simultaneously:

1.  **MCP Server**: Provides data access (employee lookup, checklists).
2.  **Agent Playground**: The web interface to chat with the agent.

We'll run these in the background.

```python
import subprocess
import time

print("Starting MCP Server...")
# Start MCP Server in the background
mcp_process = subprocess.Popen(["make", "mcp"], stdout=subprocess.PIPE, stderr=subprocess.PIPE)

print("Starting Agent Playground...")
# Start Agent Playground in the background
playground_process = subprocess.Popen(["make", "playground"], stdout=subprocess.PIPE, stderr=subprocess.PIPE)

print("Services are starting up... waiting 10 seconds...")
time.sleep(10)

print("\n🚀 **HireFlow is Ready!**")
print("Click the link below to open the Agent Playground:")
print("http://localhost:8501/?app=hireflow_agent")
```

## 4. Interact with HireFlow

1.  Open the link above (usually `http://localhost:8501/?app=hireflow_agent`).
2.  In the chat interface, try saying: **"Hi, I'm a new hire."**
3.  When asked for your email, use: **`john.doe@xyz.com`** (this is a mock employee in our database).
4.  Follow the conversation to complete your onboarding tasks!

## 5. Cleanup

When you are done, run the cell below to stop the background services.

```python
print("Stopping services...")
mcp_process.terminate()
playground_process.terminate()
print("Services stopped.")
```
